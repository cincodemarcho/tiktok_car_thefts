```python
import matplotlib
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import matplotlib.ticker as mticker
from matplotlib.patches import FancyBboxPatch
```

```python
month_map = {'Jan':1, 'Feb':2, 'Mar':3, 'Apr':4, 'May':5, 'Jun':6, 'Jul':7, 'Aug':8, 'Sep':9, 'Oct':10, 'Nov':11, 'Dec':12}

for df in [kh_thefts, mlwk_data]:
    df['month_num'] = df['month'].map(month_map)
    df['date'] = pd.to_datetime(dict(year = df['year'], month = df['month_num'], day = 1))

theft_map['pct'] = pd.to_numeric(theft_map['percentChange2019to2022'], errors = 'coerce')
```

```python
city_cols = [
    (col, str(mthrbd_data.iloc[0, col]))
    for col in range(1, mthrbd_data.shape[1], 3)
    if pd.notna(mthrbd_data.iloc[0, col])]
    
rows = []
for i in range(2, mthrbd_data.shape[0]):
    date_val = mthrbd_data.iloc[i, 0]
    if pd.isna(date_val):
        continue
    for col, city in city_cols:
        rows.append({
            'date': pd.to_datetime(date_val),
            'city': city,
            'kia_hyundai': pd.to_numeric(mthrbd_data.iloc[i, col],     errors = 'coerce'),
            'all_thefts': pd.to_numeric(mthrbd_data.iloc[i, col + 1], errors = 'coerce'),
            'pct': pd.to_numeric(mthrbd_data.iloc[i, col + 2], errors = 'coerce')})
    
mthrbd_long = pd.DataFrame(rows)
mthrbd_long['year'] = mthrbd_long['date'].dt.year
```

## Visualization 1 - Pie small multiples: national share by year

```python
yearly = (kh_thefts.groupby('year').agg(kia_h = ('countKiaHyundaiThefts','sum'),
                                        other = ('countOtherThefts','sum')).reset_index())

yearly['pct'] = yearly['kia_h'] / (yearly['kia_h'] + yearly['other'])
 
fig, axes = plt.subplots(1, 4, figsize = (15, 5))
 
for ax, (_, row) in zip(axes, yearly.iterrows()):
    vals  = [row['other'], row['kia_h']]
    kpct  = row['pct'] * 100
    ax.pie(vals, startangle = 90, wedgeprops = dict(edgecolor = 'white'), counterclock = False)  
    ax.text(0.4,  0.5, f'{kpct:.1f}%', ha = 'center', va = 'center', color = "orange", fontsize = 18) 
    ax.text(0, -0.18, str(int(row['year'])), ha = 'center', va = 'center', fontsize = 15, fontweight = 'bold')

legend_patches = [mpatches.Patch(label = 'All Other Vehicles'), mpatches.Patch(label = 'Kia / Hyundai', color = "orange")] 
fig.legend(handles = legend_patches, loc = 'lower center', bbox_to_anchor = (0.5, -0.04), ncol = 2) 
fig.suptitle('Kia/Hyundai Share of All Vehicle Thefts, 2019 - 2022', fontweight = 'bold') 

fig.savefig('Vis 1 - Pie chart')
plt.show()
```

## Visualization 2 - Donut: Milwaukee April 2021 peak

```python
peak = mlwk_data[(mlwk_data['month'] == 'Apr') & (mlwk_data['year'] == 2021)].iloc[0]
vals  = [peak['countOtherThefts'], peak['countKiaHyundaiThefts']]
total = sum(vals)
pcts  = [v / total * 100 for v in vals]
 
fig, ax = plt.subplots(figsize = (10, 5))
wedges, _ = ax.pie(vals, startangle = 90, wedgeprops = dict(width = 0.50), counterclock = False) 
 
for wedge, pct in zip(wedges, pcts):
    angle = (wedge.theta2 + wedge.theta1) / 2
    x = 0.72 * np.cos(np.radians(angle))
    y = 0.72 * np.sin(np.radians(angle))
    ax.text(x, y, f'{pct:.1f}%', ha = 'center', va = 'center', fontsize = 15, fontweight = 'bold', color = "white")  
 
ax.text(0,  0.08, f'{total:,}', ha = 'center', va = 'center', fontsize = 20, fontweight = 'bold')  
ax.text(0, -0.18, 'Total Thefts', ha = 'center', va = 'center', fontsize = 15)   
 
legend_patches = [mpatches.Patch(label = 'All Other Vehicles'), mpatches.Patch(label = 'Kia / Hyundai', color = "orange")] 
ax.legend(handles = legend_patches, loc = 'lower center', bbox_to_anchor = (0.5, -0.08), ncol = 2)
ax.set_title("April 2021: 71.4% of ALL Car Thefts Were Kia or Hyundai", fontweight = 'bold')  

fig.savefig('Vis 2 - Donut')
plt.show()
```

## Visualization 3 - Stacked Horizontal Bar: Cities in 2022

Stacked Bars with Categorical Data

```python
c2 = (kh_thefts[kh_thefts['year'] == 2022].groupby('city')
      .agg(kia_h = ('countKiaHyundaiThefts','sum'), other = ('countOtherThefts', 'sum')).reset_index().sort_values('kia_h'))
 
fig, ax = plt.subplots(figsize = (10, 5))
bars_o = ax.barh(c2['city'], c2['other'], label = 'All Other Vehicles')  #, height = 0.65)  
bars_k = ax.barh(c2['city'], c2['kia_h'], left = c2['other'], label = 'Kia / Hyundai')  #, height = 0.65)  
 
for bar, val in zip(bars_k, c2['kia_h']):
    ax.text(bar.get_x() + bar.get_width() + 50,
            bar.get_y() + bar.get_height() / 2,
            f'{val:,}', va = 'center', fontsize = 8) 
 
ax.xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'{int(x):,}'))
ax.set_xlabel('Total Thefts (2022)')  #, fontsize = 10)
ax.legend(loc = 'lower right')
ax.set_title('Kia/Hyundai vs. All Other Thefts by City (2022)', fontweight = 'bold')  
ax.grid(axis = 'x')

fig.savefig('Vis 3 - Stacked bar')
plt.show()
```

## Visaulization 4 - Treemap: car theft % change by city (manual matplotlib)

```python
def treemap_rects(values, x, y, w, h):
    rects = []
    _tm_fill(list(values), list(range(len(values))), x, y, w, h, sum(values), rects)
    return rects


def _tm_fill(vals, idxs, x, y, w, h, total, rects):
    if not vals:
        return
    if len(vals) == 1:
        rects.append((idxs[0], x, y, w, h))
        return
    cumsum = np.cumsum(vals)
    split = max(1, min(np.searchsorted(cumsum, total / 2), len(vals) - 1))
    l_vals, r_vals = vals[:split], vals[split:]
    l_idxs, r_idxs = idxs[:split], idxs[split:]
    l_tot,  r_tot  = sum(l_vals), sum(r_vals)
    if w >= h:
        w1 = w * l_tot / total
        _tm_fill(l_vals, l_idxs, x,      y, w1,     h, l_tot, rects)
        _tm_fill(r_vals, r_idxs, x + w1, y, w - w1, h, r_tot, rects)
    else:
        h1 = h * l_tot / total
        _tm_fill(l_vals, l_idxs, x, y,      w, h1,     l_tot, rects)
        _tm_fill(r_vals, r_idxs, x, y + h1, w, h - h1, r_tot, rects)


def cat_color(p):
    if p >= 2.0: return '#7B1D1D'
    if p >= 1.0: return 'red'
    if p >= 0.5: return '#F4A261'
    if p >= 0.0: return '#A8C5DA'
    return 'blue'
```

```python
top40 = (theft_map.dropna(subset=['pct', 'countCarThefts2022'])
         .assign(count22 = lambda d: pd.to_numeric(d['countCarThefts2022'], errors = 'coerce')).dropna(subset = ['count22'])
         .query('count22 > 100').sort_values('count22', ascending = False).head(40).reset_index(drop = True))
 
top40['color'] = top40['pct'].apply(cat_color)
sizes = top40['count22'].values
rects = treemap_rects(sizes.tolist(), 0, 0, 1.0, 1.0)

fig, ax = plt.subplots(figsize = (15, 10))
ax.set_xlim(0, 1); ax.set_ylim(0, 1)
ax.axis('off')
#ax.set_facecolor("#FAFAFA")
 
for (idx, rx, ry, rw, rh) in rects:
    row = top40.iloc[idx]
    patch = FancyBboxPatch((rx + 0.003, ry + 0.003), rw - 0.006, rh - 0.006, boxstyle = "round,pad=0.004", linewidth = 1.5, 
                           facecolor = row['color'], edgecolor = 'white')
    ax.add_patch(patch)
    if rw > 0.05 and rh > 0.04:
        name = str(row['geo_name']).replace(' PD', '').replace(' SO', '')
        pct_str = f"{row['pct']*100:.0f}%"
        fs = max(5.5, min(9, rw * 65))
        ax.text(rx + rw/2, ry + rh/2 + rh * 0.08, name, ha = 'center', va = 'center', fontsize = fs, fontweight = 'bold' , color = 'white')  
        ax.text(rx + rw/2, ry + rh/2 - rh * 0.12, pct_str, ha = 'center', va = 'center', fontsize = fs * 0.9, alpha = 0.92, color = 'white')  
        
legend_items = [
    ('>200% increase', '#7B1D1D'),
    ('100-200% increase', "red"),
    ('50-100% increase', '#F4A261'),
    ('0-50% increase', '#A8C5DA'),
    ('Decrease', "blue")]

patches = [mpatches.Patch(color = c, label = l) for l, c in legend_items]
ax.legend(handles = patches, loc = 'lower center', bbox_to_anchor = (0.5, -0.07), ncol = 5, frameon = False, fontsize = 9)
ax.set_title('% Change in Car Thefts by City (2019 to 2022)\n'
             'Color = % change since 2019  |  Top 40 cities', fontweight = 'bold') 

fig.savefig('Vis 4 - Tree map')
plt.show()
```

## Visualizaiton 5 - Area Chart: Milwaukee % share over time

```python
milw_s = mlwk_data.sort_values('date')
 
fig, ax = plt.subplots(figsize = (10, 5))

ax.fill_between(milw_s['date'], milw_s['percentKiaHyundai'] * 100, alpha = 0.25)
ax.plot(milw_s['date'], milw_s['percentKiaHyundai'] * 100, linewidth = 2)

peak_row = milw_s.loc[milw_s['percentKiaHyundai'].idxmax()]
ax.scatter(peak_row['date'], peak_row['percentKiaHyundai'] * 100, color = 'red', zorder = 5, s = 80)
ax.annotate('April 2021: 71.4%\nPeak of the crisis',
            xy = (peak_row['date'], peak_row['percentKiaHyundai'] * 100),
            xytext = (peak_row['date'] - pd.Timedelta(days = 480),
                      peak_row['percentKiaHyundai'] * 100 - 10), fontweight = 'bold', color = 'red',
            arrowprops = dict(arrowstyle = '->', color = 'red'))  

ax.axvline(pd.Timestamp('2021-07-01'), color = 'black', linestyle = '--')  
ax.text(pd.Timestamp('2021-07-15'), 20, 'TikTok trend goes national', color = 'black')
 
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'{x:.0f}%'))
ax.set_ylim(0, 80)
ax.set_ylabel('% of All Car Thefts in Milwaukee') 
ax.set_title("Milwaukee's Kia/Hyundai Theft Crisis (2019 - 2022)", fontweight='bold')

fig.savefig('Vis 5 - Area chart')
plt.show()
```
